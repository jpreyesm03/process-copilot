```python
import pandas as pd
pd.set_option("display.max_columns", 30)
pd.set_option("display.max_rows", 45)

df = pd.read_csv("../data/processed/bpi2019_events.csv.gz")
```

```python
print(df.info())
display(df.head())
```

```python
test = df.groupby("case_purchasing_document").agg({'case_id' : 'nunique'}).sort_values(by='case_id', ascending =False).reset_index()
print("Case purchasing document with only one case_id:", test[test["case_id"] == 1].shape[0])
print("Nunique of case purchasing document:", df["case_purchasing_document"].nunique())
print("Nunique of case_id:", df["case_id"].nunique())
```
